```python
# Инициализация Spark сессии
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
spark = SparkSession.builder \
    .appName("HW2_DSAM") \
    .getOrCreate()
```

```python
# 1. Считайте файл 1.json и создайте Spark DataFrame
df = spark.read.json('1.json')
```

```python
# 2. Выведите первые 14 строк
print("Первые 14 строк:")
df.show(14)
```

```python
# 3. Выведите количество колонок
print("Количество колонок:", len(df.columns))
print("Имена колонок:", df.columns)
```

```python
# 4. Выведите статистические характеристики для всех колонок
print("Статистические характеристики:")
df.describe().show()
```

```python
# 5. Удалите указанную колонку: expiration_date
df = df.drop("expiration_date")
df.show(5)
```

```python
# 6. Создайте новую колонку: цена в $ (price * 91.0)
df = df.withColumn("price_usd", col("price") * 91.0)
df.show(5)
```

```python
# 7. Запишите полученный набор данных в файл: r_1.csv
df.write.csv("r_1.csv", header=True, mode="overwrite")
```

```python
# Проверка результатов
print("Итоговый DataFrame после изменений:")
df.show(5)
```
